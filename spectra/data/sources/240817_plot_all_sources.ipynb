```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
from spectra.sourcespectrum import SourceSpectrum

%matplotlib widget
```

```python
cwd = Path.cwd()
directories = sorted(
    [d for d in cwd.iterdir() if d.is_dir() and d.name[0] not in [".", "_"]]
)
directories
```

```python
sources = {}
for d in directories:
    sources[d.name] = SourceSpectrum(d)
    
sources
```

```python
fig, ax = plt.subplots()
for key, spectrum in sources.items():
    ax.plot(spectrum.wavelength, spectrum.power, label=key)
ax.grid()
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Normalized Power')
ax.legend();
```
