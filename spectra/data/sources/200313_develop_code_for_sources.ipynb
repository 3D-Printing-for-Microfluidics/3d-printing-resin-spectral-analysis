```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
%matplotlib widget
```

```python
cwd = Path.cwd()
directories = sorted(
    [d for d in cwd.iterdir() if d.is_dir() and d.name[0] not in [".", "_"]]
)
directories
```

```python
class SourceSpectrum:
    """Optical spectrum data.
    """
    def __init__(self, directory):
        self._data = np.loadtxt(directory / 'measured_spectrum_normalized.csv', delimiter=',')
        self.wavelength = self._data[:, 0]
        self.power = self._data[:, 1]
```

```python
sources = {}
for d in directories:
    sources[d.name] = SourceSpectrum(d)
    
sources
```

```python
fig, ax = plt.subplots()
for key, spectrum in sources.items():
    ax.plot(spectrum.wavelength, spectrum.power, label=key)
ax.grid()
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Normalized Power')
ax.legend();
```
