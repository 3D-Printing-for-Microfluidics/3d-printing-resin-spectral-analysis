# Code

```python
def get_raw_data(data_path, directory_name, skiprows=14):
    
    assert isinstance(data_path, Path)
    
    raw_data_dir = data_path / directory_name / "raw_data"
    raw_data_files = [f for f in raw_data_dir.glob("*.txt")]
    
    data = {}
    for f in raw_data_files:
        data[f] = np.loadtxt(f, skiprows=skiprows)

    return data

    
def plot_data(data, normalize=False):
    
    assert isinstance(data, dict)
    
    fig, ax = plt.subplots(figsize=(10,8))
    for k,d in data.items():
        max_d = np.max(d[:,1]) if normalize else 1
        ax.plot(d[:,0], d[:,1]/max_d, label=k.name)
#     ax.set_xlim(300, 540)
    ax.set_xlabel('Wavelength (nm)')
    ylabel = 'Normalized Power' if normalize else 'Power'
    ax.set_ylabel(ylabel)
    ax.legend(loc='upper left')
    ax.grid()
    
    return fig, ax
    
```
