# MR1 Visitech 2020-03-25

```python
raw_data_dir = data_path / "MR1_385nm_Visitech-2020-03-25" / "raw_data"
raw_data_dir
```

```python
raw_data_files = [f for f in raw_data_dir.glob("*.txt")]
raw_data_files
```

```python
data = {}
for f in raw_data_files:
    data[f] = np.loadtxt(f, skiprows=14)
```

```python
fig, ax = plt.subplots(figsize=(10,8))
for k,d in data.items():
    ax.plot(d[:,0], d[:,1], label=k.name)
ax.set_xlim(300, 540)
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Power')
ax.set_title('MR1 Visitech 2020-03-25')
ax.legend()
ax.grid();
```

```python
fig, ax = plt.subplots(figsize=(10,8))
for k,d in data.items():
    ax.plot(d[:,0], d[:,1]/np.max(d[:,1]), label=k.name)
ax.set_xlim(300, 540)
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Normalized Power')
ax.set_title('MR1 Visitech 2020-03-25')
ax.legend()
ax.grid();
```
