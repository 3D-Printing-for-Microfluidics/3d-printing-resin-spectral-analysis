```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
from spectra.absorberspectrum import AbsorberSpectrum
%matplotlib widget
```

```python
cwd = Path.cwd()
directories = sorted(
    [d for d in cwd.iterdir() if d.is_dir() and d.name[0] not in [".", "_"]]
)
directories
```

```python
absorbers = {}
for d in directories:
    absorbers[d.name] = AbsorberSpectrum(d)
    
absorbers
```

```python
fig, ax = plt.subplots()
for key, spectrum in absorbers.items():
    ax.plot(spectrum.wavelength, spectrum.molar_absorptivity, label=key)
ax.grid()
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Molar Absorptivity (L $\cdot$ mol$^{-1} \cdot$cm$^{-1}$)')
ax.legend();
```
