```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
# %matplotlib widget
```

# Purpose

Make a known source that is easy to reason about and use for exploration and testing. 

## Create data

```python
# Source wavelength min and max
source_min_wavelength = 360
source_max_wavelength = 370

# Wavelength range parameters
# range_min_wavelength = 320
# range_max_wavelength = 400
range_min_wavelength = 355
range_max_wavelength = 375

num_pnts = (range_max_wavelength - range_min_wavelength) * 20 + 1

source_wavelengths = np.linspace(range_min_wavelength, range_max_wavelength, num_pnts)

source_normalized_power = np.zeros(len(source_wavelengths))

source_normalized_power[np.logical_and(
    source_wavelengths >= source_min_wavelength, 
    source_wavelengths <= source_max_wavelength)] = 1
```

## Plot

```python
fig, ax = plt.subplots()
ax.plot(source_wavelengths, source_normalized_power)
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Normalized power');
plt.savefig('spectrum.png')
```

## Save to file

```python
source = np.stack((source_wavelengths,source_normalized_power)).T

header_text = f"Ideal uniform spectrum from {source_min_wavelength} nm to {source_max_wavelength} nm"

np.savetxt('measured_spectrum_normalized.csv', source, delimiter=',', header=header_text)
```
