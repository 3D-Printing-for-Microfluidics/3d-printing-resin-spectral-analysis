```python
avobenzone_data = {}
for key, value in ab.absorbers.items():
    if key.startswith('avobenzone'):
        avobenzone_data[key] = value
        
avobenzone_data
```

```python
fig, ax = plt.subplots()
for key, value in avobenzone_data.items():
    ax.plot(value.wavelength, value.molar_absorptivity, label=key)
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Molar Absorptivity (L $\cdot$ mol$^{-1} \cdot$cm$^{-1}$)')
ax.grid()
ax.legend();
```
