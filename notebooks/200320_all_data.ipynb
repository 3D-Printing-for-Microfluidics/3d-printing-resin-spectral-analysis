```python
import json
from pathlib import Path
import sys

import numpy as np
import matplotlib.pyplot as plt
%matplotlib widget

import spectra.data as data
import spectra
```

# Absorbers

```python
data.absorbers
```

```python
fig, ax = plt.subplots(figsize=(10,8))
for k,d in data.absorbers.items():
    ax.plot(d.wavelength_absorbance, d.absorbance, label=k)
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Absorbance')
ax.set_title('Absorbers')
ax.legend()
ax.grid();
```

```python
fig, ax = plt.subplots(figsize=(10,8))
for k,d in data.absorbers.items():
    ax.plot(d.wavelength, d.molar_absorptivity, label=k)
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Molar Absorptivity (L $\cdot$ mol$^{-1} \cdot$cm$^{-1}$)')
ax.set_title('Absorbers')
ax.legend()
ax.grid();
```

# Sources

```python
data.sources
```

## All

```python
fig, ax = plt.subplots(figsize=(10,8))
for k,d in data.sources.items():
    ax.plot(d.wavelength, d.power, label=k)
ax.set_xlim(300, 540)
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Normalized Power')
ax.set_title('Sources')
ax.legend()
ax.grid();
```

## HR3.3 with and without tray

```python
temp_dict = {k:data.sources[k] for k in data.sources if k.startswith('HR3.3')}

fig, ax = plt.subplots(figsize=(10,8))
for k,d in temp_dict.items():
    ax.plot(d.wavelength, d.power, label=k)
# ax.set_xlim(300, 540)
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Normalized Power')
ax.set_title('Sources')
ax.legend()
ax.grid();
```
