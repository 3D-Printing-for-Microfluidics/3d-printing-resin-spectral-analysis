```python
fig, ax = plt.subplots()
for key, spectrum in sources.items():
    ax.plot(spectrum.wavelength, spectrum.power, label=key)
ax.grid()
ax.set_xlabel('Wavelength (nm)')
ax.set_ylabel('Normalized Power')
ax.legend();
```
